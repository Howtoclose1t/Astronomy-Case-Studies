```python
import importlib.util
from pathlib import Path


inventory_file = Path(
    r"D:\AST\04_JADES_Catalog_Matching"
    r"\src\dr45_inventory.py"
)

inventory_spec = (
    importlib.util.spec_from_file_location(
        "dr45_inventory",
        inventory_file,
    )
)

dr45_inventory = (
    importlib.util.module_from_spec(
        inventory_spec
    )
)

inventory_spec.loader.exec_module(
    dr45_inventory
)

inventory_results = (
    dr45_inventory.run_inventory()
)

display(
    inventory_results[
        "hdu_summary"
    ]
)

display(
    inventory_results[
        "core_columns"
    ]
)

display(
    inventory_results[
        "filter_coverage"
    ]
)

print(
    "Full column inventory:",
    inventory_results[
        "column_inventory_path"
    ],
)
```

```python
from pathlib import Path
import subprocess
import sys

import pandas as pd
from IPython.display import display


TASK4_ROOT = Path(
    r"D:\AST\04_JADES_Catalog_Matching"
)

completed_process = subprocess.run(
    [
        sys.executable,
        "-m",
        "src.dr45_matching",
    ],
    cwd=TASK4_ROOT,
    text=True,
    capture_output=True,
)

if completed_process.returncode != 0:
    print(completed_process.stdout)
    raise RuntimeError(
        completed_process.stderr
    )

match_summary = pd.read_csv(
    TASK4_ROOT
    / "outputs"
    / "metrics"
    / "dr45_match_summary.csv"
)

method_summary = pd.read_csv(
    TASK4_ROOT
    / "outputs"
    / "metrics"
    / "dr45_match_method_summary.csv"
)

display(match_summary)
display(method_summary)
```

```python
from pathlib import Path
import subprocess
import sys

import pandas as pd
from IPython.display import display


TASK4_ROOT = Path(
    r"D:\AST\04_JADES_Catalog_Matching"
)

completed_process = subprocess.run(
    [
        sys.executable,
        "-m",
        "src.dr45_photometry",
    ],
    cwd=TASK4_ROOT,
    text=True,
    capture_output=True,
)

if completed_process.returncode != 0:
    print(completed_process.stdout)
    raise RuntimeError(
        completed_process.stderr
    )

photometry_summary = pd.read_csv(
    TASK4_ROOT
    / "outputs"
    / "metrics"
    / "dr45_photometry_summary.csv"
)

filter_coverage = pd.read_csv(
    TASK4_ROOT
    / "outputs"
    / "metrics"
    / "dr45_matched_filter_coverage.csv"
)

coverage_table = (
    filter_coverage
    .pivot(
        index="filter",
        columns="field",
        values="valid_fraction",
    )
    .reset_index()
)

display(photometry_summary)
display(coverage_table)
```
